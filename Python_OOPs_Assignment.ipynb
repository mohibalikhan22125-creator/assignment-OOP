**Q1:** Aik `Student` class banayein jismein `name` aur `roll_number` private variables (`__`) hon. In variables ko access aur modify karne ke liye getter aur setter methods likhein.

```python
# Write your code here for Q1
class Student:

    def __init__(self, name, roll_number):
        self.__name = name
        self.__roll_number = roll_number

    # Getter for name
    def get_name(self):
        return self.__name

    # Setter for name
    def set_name(self, name):
        self.__name = name

    # Getter for roll_number
    def get_roll_number(self):
        return self.__roll_number

    # Setter for roll_number
    def set_roll_number(self, roll_number):
        self.__roll_number = roll_number


# Object banana
student1 = Student("Mohib", 101)

# Getter se values lena
print(student1.get_name())
print(student1.get_roll_number())

# Setter se values change karna
student1.set_name("Ali")
student1.set_roll_number(102)

print(student1.get_name())
print(student1.get_roll_number())
```

**Q2:** Aik `BankAccount` class banayein jismein `__balance` private variable ho. Aik `deposit()` aur aik `withdraw()` method banayein, check karein ke withdraw karte waqt balance zero se kam na ho.

```python
class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount
        print("Deposit successful!")

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
            print("Withdraw successful!")
        else:
            print("Insufficient balance!")

    def show_balance(self):
        print("Balance:", self.__balance)


# Object banana
account = BankAccount(10000)

# Balance check
account.show_balance()

# Deposit
account.deposit(5000)
account.show_balance()

# Withdraw
account.withdraw(3000)
account.show_balance()

# Zyada amount withdraw karna
account.withdraw(15000)
account.show_balance()
```

**Q3:** Aik class banayein `Employee` jismein salary private ho. Agar koi salary ko minus (negative) value set karne ki koshish kare, toh setter method 'Invalid Salary' ka error message show kare.

```python
# Write your code here for Q3
class Employee:

    def __init__(self, salary):
        self.__salary = salary

    # Getter
    def get_salary(self):
        return self.__salary

    # Setter
    def set_salary(self, salary):
        if salary >= 0:
            self.__salary = salary
        else:
            print("Invalid Salary")


# Object banana
employee1 = Employee(50000)

# Salary dekhna
print(employee1.get_salary())

# Valid salary set karna
employee1.set_salary(60000)
print(employee1.get_salary())

# Negative salary set karna
employee1.set_salary(-10000)
print(employee1.get_salary())
```

**Q12:** Polymorphism function ke zariye kaise achieve hoti hai? Aik function banayein `make_sound(animal_object)` jo har tarah ke animal class ka sound method automatically call kare.

```python
# Write your code here for Q12
class Dog:

    def sound(self):
        print("Dog says: Woof Woof")


class Cat:

    def sound(self):
        print("Cat says: Meow")


class Cow:

    def sound(self):
        print("Cow says: Moo")


# Polymorphism function
def make_sound(animal_object):
    animal_object.sound()


# Objects banana
dog1 = Dog()
cat1 = Cat()
cow1 = Cow()

# Function call
make_sound(dog1)
make_sound(cat1)
make_sound(cow1)
```

**Q13:** Operator Overloading kya hoti hai? Do objects ko `+` operator ke zariye add karne ke liye `__add__` magic method ka use kaise karte hain? Code likhein.

```python
# Write your code here for Q13
class Number:

    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value


# Do objects banana
num1 = Number(10)
num2 = Number(20)

# Objects ko + se add karna
result = num1 + num2

print("Result:", result)
```

**Q15:** Kya Python default roop se Method Overloading (same method name with different arguments) support karta hai? Agar nahi, toh hum same name ke method ko multiple arguments ke sath kaise handle karte hain? Code likhein.

```python
# Write your code here for Q15
class Calculator:

    def add(self, *numbers):
        total = 0

        for number in numbers:
            total += number

        return total


calc = Calculator()

print(calc.add(10, 20))
print(calc.add(10, 20, 30))
print(calc.add(10, 20, 30, 40))
```

**Q17:** Q16 wali `Computer` class ko inherit karke do child classes `Laptop` aur `Desktop` banayein aur dono mein `process()` method ko mandatory implement karein.

```python
# Write your code here for Q17
from abc import ABC, abstractmethod


class Computer(ABC):

    @abstractmethod
    def process(self):
        pass


class Laptop(Computer):

    def process(self):
        print("Laptop is processing")


class Desktop(Computer):

    def process(self):
        print("Desktop is processing")


# Objects banana
laptop1 = Laptop()
desktop1 = Desktop()

# Process method call karna
laptop1.process()
desktop1.process()
```

**Q18:** Aik abstract class `Car` banayein jisme aik concrete (normal) method `start_engine()` ho aur aik abstract method `drive()` ho. Child class bana kar dono ko test karein.

```python
# Write your code here for Q18
from abc import ABC, abstractmethod


class Car(ABC):

    # Concrete / normal method
    def start_engine(self):
        print("Engine started")

    # Abstract method
    @abstractmethod
    def drive(self):
        pass


class Toyota(Car):

    # Abstract method ko implement karna zaroori hai
    def drive(self):
        print("Toyota is driving")


# Child class ka object
car1 = Toyota()

# Dono methods test karna
car1.start_engine()
car1.drive()
```

**Q19:** Abstraction aur Encapsulation ke darmiyan kya wazeh farq (main difference) hai? Dono ke concepts ko short statements mei define karein.

```python
# Write your code here for Q19
from abc import ABC, abstractmethod

# Abstraction
class Car(ABC):

    @abstractmethod
    def drive(self):
        pass


class Toyota(Car):

    def drive(self):
        print("Car is driving")


car = Toyota()
car.drive()


# Encapsulation
class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    def show_balance(self):
        print("Balance:", self.__balance)


account = BankAccount(50000)
account.show_balance()
```

**Q20:** Aik real-world scenario (jaise Payment System) design karein jahan `Payment` aik abstract class ho aur `CreditCardPayment` aur `UPIPayment` uski subclasses hon.

```python
# Write your code here for Q20
from abc import ABC, abstractmethod


# Abstract class
class Payment(ABC):

    @abstractmethod
    def pay(self, amount):
        pass


# Credit Card Payment
class CreditCardPayment(Payment):

    def pay(self, amount):
        print("Credit Card se payment:", amount)


# UPI Payment
class UPIPayment(Payment):

    def pay(self, amount):
        print("UPI se payment:", amount)


# Objects banana
payment1 = CreditCardPayment()
payment2 = UPIPayment()

# Payments karna
payment1.pay(5000)
payment2.pay(3000)
```
